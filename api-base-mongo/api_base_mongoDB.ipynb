```python
import requests

def appel_api_base_add_data(url = URL_LOCAL, datas):
    url = url + "/add_data"
    response = requests.post(url,json=datas)
    print(f"code http: "+str(response.status_code))
    print(response.text)

appel_api_base_add_data(repositories)
```

```python

import requests
    
def appel_api_base_show_data(url = URL_LOCAL,page = 1,page_size = 10000):
    url = url +"/show_data?page="+str(page)+"&page_size="+str(page_size)
    response = requests.get(url)
    print(f"code http: "+str(response.status_code))
    return (response.text)
    
reponse = appel_api_base_show_data()
print(reponse)
```
